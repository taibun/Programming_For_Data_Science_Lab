### Problem 1. Write a program that inputs two numbers and displays their sum, difference, product, andquotient. Convert the inputs to numeric values before calculating.

```python
num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))

print("Sum:", num1 + num2)
print("Difference:", num1 - num2)
print("Product:", num1 * num2)

if num2 != 0:
    print("Quotient:", num1 / num2)
else:
    print("Quotient: Cannot divide by zero")
```

### Problem 3. Input a temperature in Celsius and convert it to Fahrenheit and Kelvin using the formulas:
#### F = (C × 9/5) + 32, K = C + 273.15

```python
celsius = float(input("Enter temperature in Celsius: "))

fahrenheit = (celsius * 9 / 5) + 32
kelvin = celsius + 273.15

print("Celsius: ", celsius, "°C")
print("Fahrenheit: ", fahrenheit, "°F")
print("Kelvin: ", kelvin, "K")
```

### Problem 4. Input a student’s name and five course marks. Calculate the total, average, highest mark,lowest mark, and number of passed courses. Display all results. A mark of 50 or higher is considered a pass.

```python
name = input("Enter student name: ")

data_science = float(input("Enter Data Science marks: "))
data_processing = float(input("Enter Data Processing marks: "))
statistics = float(input("Enter Statistics marks: "))
python_programming = float(input("Enter Python Programming marks: "))
machine_learning = float(input("Enter Machine Learning marks: "))

total = 0
passed = 0
highest = 0
lowest = 100

for mark in [data_science, data_processing, statistics, python_programming, machine_learning]:
    total += mark

    if mark >= 50:
        passed += 1

    if mark > highest:
        highest = mark

    if mark < lowest:
        lowest = mark

average = total / 5

print()
print("Student Name:", name)
print("Total Marks:", total)
print("Average Marks:", average)
print("Highest Mark:", highest)
print("Lowest Mark:", lowest)
print("Passed Courses:", passed)
```
