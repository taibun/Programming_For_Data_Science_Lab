### Problem 1 – Course Marks. Create a list containing the marks of five courses. Display the total,average, highest mark, and lowest mark. Count the number of passed courses, using 50 as the pass mark.

```python
marks = [94, 60, 55, 80, 82]

total = 0
passed = 0
highest = marks[0]
lowest = marks[0]

for mark in marks:
    total += mark

    if mark >= 50:
        passed += 1

    if mark > highest:
        highest = mark

    if mark < lowest:
        lowest = mark

average = total / len(marks)

print("Marks:", marks)
print("Total:", total)
print("Average:", average)
print("Highest Mark:", highest)
print("Lowest Mark:", lowest)
print("Passed Courses:", passed)
```

### Problem 2 – Weekend Temperatures. Create a tuple containing six temperature readings: three for Saturday and three for Sunday. Use a for loop to display each reading and calculate the highest and lowest temperatures.

```python
temperatures = (30, 32, 29, 31, 34, 28)

highest = temperatures[0]
lowest = temperatures[0]

for i in range(len(temperatures)):
    if i < 3:
        print("Saturday:", temperatures[i])
    else:
        print("Sunday:", temperatures[i])

    if temperatures[i] > highest:
        highest = temperatures[i]

    if temperatures[i] < lowest:
        lowest = temperatures[i]

print("Highest Temperature:", highest)
print("Lowest Temperature:", lowest)
```

### Problem 3 – Word Analyzer. Ask the user to enter a word or short sentence. Display the numberof characters, vowels (a, e, i, o, u), and spaces. Use a for loop to examine each character.

```python
text = input("Enter a word or short sentence: ")

characters = 0
vowels = 0
spaces = 0

for ch in text:
    characters += 1

    if ch.lower() in "aeiou":
        vowels += 1

    if ch == " ":
        spaces += 1

print("Characters:", characters)
print("Vowels:", vowels)
print("Spaces:", spaces)
```

## Problem 4 – Grade Calculator

Write a function named `calculate_grade(marks)` that receives a list of marks, calculates the average, and returns the corresponding grade.

| Average | Grade |
|---|---|
| 80 or higher | A |
| 70 to below 80 | B |
| 60 to below 70 | C |
| 50 to below 60 | D |
| Below 50 | F |

Call the function using a list of five marks.

```python
def calculate_grade(marks):
    average = sum(marks) / len(marks)

    if average >= 80:
        return "A"
    elif average >= 70:
        return "B"
    elif average >= 60:
        return "C"
    elif average >= 50:
        return "D"
    else:
        return "F"


marks = [85, 75, 65, 80, 70]

print("Average:", sum(marks) / len(marks))
print("Grade:", calculate_grade(marks))
```

## Problem 5 – Shopping Cart Processor

Create a list of dictionaries representing four products. Each dictionary must contain a product name, price, and quantity.

Use a loop to calculate the total cost. Write a function named `apply_discount(total)` that applies a 10% discount when the total is at least 2,000. Otherwise, return the original total.

```python
products = [
    {"name": "Keyboard", "price": 1200, "quantity": 1},
    {"name": "Mouse", "price": 600, "quantity": 2},
    {"name": "USB Cable", "price": 200, "quantity": 2},
    {"name": "Notebook", "price": 100, "quantity": 1}
]

total = 0

for product in products:
    total += product["price"] * product["quantity"]


def apply_discount(total):
    if total >= 2000:
        return total * 0.90
    else:
        return total


print("Total Cost:", total)
print("Final Cost:", apply_discount(total))
```

## Problem 6 – Countdown

Write a program using a `while` loop that asks the user for a non-negative integer and prints a countdown from that number to zero.

```python
num = int(input("Enter a non-negative integer: "))

while num >= 0:
    print(num)
    num -= 1
```

## Problem 7 – Student Record Analyzer

Using the student records below, display each student's name and marks. Count the students who passed (marks >= 50), calculate the class average, and identify the student with the highest mark.

```python
students = [
    {"name": "Rahim", "department": "CSE", "marks": 85, "attendance": 92},
    {"name": "Karim", "department": "CSE", "marks": 48, "attendance": 76},
    {"name": "Sara", "department": "DS", "marks": 91, "attendance": 95},
    {"name": "Nadia", "department": "DS", "marks": 67, "attendance": 84},
    {"name": "Hasan", "department": "CSE", "marks": 73, "attendance": 78}
]

total = 0
passed = 0
highest = students[0]

for student in students:
    print("Name:", student["name"], "Marks:", student["marks"])

    total += student["marks"]

    if student["marks"] >= 50:
        passed += 1

    if student["marks"] > highest["marks"]:
        highest = student

average = total / len(students)

print("Passed Students:", passed)
print("Class Average:", average)
print("Highest Mark:", highest["marks"])
print("Top Student:", highest["name"])
```

## Problem 8 – Product Inventory Summary

Using the product records below, calculate each product's line total using price * quantity. Calculate the total inventory value and identify the product with the highest line total.

```python
products = [
    {"name": "Notebook", "price": 120, "quantity": 3},
    {"name": "Pen", "price": 20, "quantity": 5},
    {"name": "USB Drive", "price": 850, "quantity": 2},
    {"name": "Mouse", "price": 650, "quantity": 1}
]

total = 0
highest = 0
highest_product = ""

for product in products:
    line_total = product["price"] * product["quantity"]

    print(product["name"], "Line Total:", line_total)

    total += line_total

    if line_total > highest:
        highest = line_total
        highest_product = product["name"]

print("Total Inventory Value:", total)
print("Highest Line Total:", highest)
print("Highest Value Product:", highest_product)
```
